# PPE YOLOv8 Training — Google Colab

Trains a YOLOv8 model on your PPE dataset and saves  to Google Drive.

**Runtime:** GPU (T4 or better) — Runtime → Change runtime type → GPU


```python
# 1. Install Ultralytics
!pip install ultralytics -q
```

```python
# 2. Mount Google Drive
from google.colab import drive
drive.mount("/content/drive")
```

```python
# 3. Set paths — edit DRIVE_DIR to your folder
import os
DRIVE_DIR  = "/content/drive/MyDrive/ppe-poc"
DATASET    = f"{DRIVE_DIR}/dataset/data.yaml"  # your dataset YAML
OUTPUT_DIR = DRIVE_DIR
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
# 4. Train
from ultralytics import YOLO
model = YOLO("yolov8n.pt")   # change to yolov8s.pt for more accuracy

results = model.train(
    data=DATASET,
    epochs=100,
    imgsz=640,
    batch=16,
    project=OUTPUT_DIR,
    name="ppe_run",
    exist_ok=True,
)

print(f"Best weights: {results.save_dir}/weights/best.pt")
```

```python
# 5. Save ppe.pt to Drive root
import shutil, pathlib

best = pathlib.Path(results.save_dir) / "weights" / "best.pt"
dest = pathlib.Path(DRIVE_DIR) / "ppe.pt"
shutil.copy(best, dest)
print(f"Saved → {dest}")
print("Drop this file into the repo root, then run: python scripts/export.py")
```
